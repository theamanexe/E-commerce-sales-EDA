# E-Commerce Sales — Exploratory Data Analysis (EDA)

This project explores an e-commerce sales dataset to understand:
- the structure and quality of the data
- product category and regional performance
- payment methods
- customer ratings and delivery time
- revenue distribution
- sales trends over time
- relationships between numerical variables

**Goal:** extract useful patterns and business insights from the dataset using Python, Pandas, Matplotlib, and Seaborn.

```python
yearly_revenue = df.groupby("year")["revenue"].sum()

plt.figure(figsize=(10, 5))
plt.plot(yearly_revenue.index, yearly_revenue.values, marker="o")
plt.title("Yearly Revenue Trend")
plt.xlabel("Year")
plt.ylabel("Revenue")
plt.grid()
plt.show()
```

```python
monthly_revenue = df.groupby("month")["revenue"].sum().sort_index()

plt.figure(figsize=(10, 5))
plt.plot(monthly_revenue.index, monthly_revenue.values, marker="o")
plt.title("Monthly Revenue Trend")
plt.xlabel("Month")
plt.ylabel("Revenue")
plt.xticks(range(1, 13))
plt.grid()
plt.show()
```

```python
df.groupby("product_category").agg(
    orders=("order_id", "count"),
    total_revenue=("revenue", "sum"),
    average_revenue=("revenue", "mean")
)
```

```python
sns.countplot(data=df, x="product_category")
plt.title("Orders by Product Category")
plt.show()
```

```python
revenue_by_category = df.groupby("product_category")["revenue"].sum()

revenue_by_category.plot(kind="bar")
plt.title("Revenue by Product Category")
plt.xlabel("Product Category")
plt.ylabel("Revenue")
plt.show()
```

```python
sns.histplot(data=df, x="revenue", kde=True)
plt.title("Revenue Distribution")
plt.show()
```

```python
sns.scatterplot(data=df, x="quantity", y="revenue")
plt.title("Quantity vs Revenue")
plt.show()
```

```python
sns.heatmap(df.corr(numeric_only=True), annot=True)
plt.title("Correlation Heatmap")
plt.show()
```

## Key Insights

- Electronics had the highest number of orders among the product categories, while Beauty had the lowest number of orders.
- Electronics generated the highest total revenue among the product categories.
- Clothing was the second-highest product category by number of orders.
- The West region generated the highest total revenue among the regions.
- Card was the payment method with the highest total revenue.
- The analysis of customer ratings showed differences in average ratings across product categories.
- Average delivery time varied across regions.
- Revenue showed variation across different years and months, indicating changes in sales performance over time.
- The revenue distribution showed that revenue values were not evenly distributed across all orders.
- The correlation analysis helped identify relationships between the numerical variables, but correlation does not imply causation.

## Conclusion

This exploratory data analysis examined an e-commerce sales dataset to understand
sales performance, product categories, regions, payment methods, customer ratings,
delivery time, and revenue trends.

The analysis showed that Electronics was the strongest product category in terms
of orders and total revenue. Regional and payment-method analysis also revealed
differences in revenue performance. Time-based analysis helped identify changes
in revenue across years and months, while correlation analysis provided insight
into relationships between numerical variables.

Overall, the EDA helped identify important patterns and trends in the e-commerce
dataset and provided a foundation for further analysis and business decision-making.
